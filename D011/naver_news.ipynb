```python
import requests
from bs4 import BeautifulSoup

url = 'https://news.naver.com/section/103'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                   '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

res = requests.get(url, headers=headers)
res.raise_for_status()  # 요청 실패 시 에러 발생시키기

soup = BeautifulSoup(res.text, 'html.parser')  # 뷰티풀숩 텍스트로 파싱

selector = 'ul[id^="_SECTION_HEADLINE_LIST"]'
headlines = soup.select_one(selector)

for i, li in enumerate(headlines.select('li'), start=1):
    title = li.select_one('div > div > a').get_text(strip=True)
    
    # 언론사명 (없는 기사도 있을 수 있어 예외 처리)
    press_tag = li.select_one('.sa_text_press')
    press = press_tag.get_text(strip=True) if press_tag else '언론사 정보 없음'

    print(f"{i}. [{press}] {title}")
```

```python
import requests
from bs4 import BeautifulSoup

url = 'https://news.naver.com/section/103'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                   '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

res = requests.get(url, headers=headers)
res.raise_for_status()

soup = BeautifulSoup(res.text, 'html.parser')

selector = 'ul[id^="_SECTION_HEADLINE_LIST"]'
headlines = soup.select_one(selector)

for i, li in enumerate(headlines.select('li'), start=1):
    title = li.select_one('.sa_text_title').get_text(strip=True)
    press = li.select_one('.sa_text_press').get_text(strip=True)
    print(f"{i}. {title} [{press}]")
```
