```python
import duckdb 
con = duckdb.connect()

with open("project_name.sql", encoding="utf-8") as f:
    script = f.read()

# 세미콜론(;) 기준으로 문장을 나눠서 하나의 연결(con)에서 순서대로 실행
statements = [s.strip() for s in script.split(";") if s.strip()]
for stmt in statements:
    con.execute(stmt)

print("모든 문장 실행 완료!")
```
